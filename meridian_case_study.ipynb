## Join the dataframes

profile + transactions

```python
profiled_transactions_df = pd.merge(
    profile_df,
    transcript_df,
    left_on='member_id',
    right_on='person',
    how='outer'
)

profiled_transactions_df.head()
```

```python
fully_joined_df = pd.merge(
    profiled_transactions_df.assign(
        **{'offer id': profiled_transactions_df['value'].apply(
            lambda x: x.get('offer id') if isinstance(x, dict) else None
        )}
    ),
    portfolio_df,
    on='offer id',
    how='outer'
)

fully_joined_df.head(50)
```
